```python
import gradio as gr
from   gradio import themes
def get_all_subclasses(cls):
    """获取一个类的所有子类（包括间接子类）"""
    all_subclasses = []
    
    for subclass in cls.__subclasses__():
        all_subclasses.append(subclass)
        all_subclasses.extend(get_all_subclasses(subclass))
    
    return all_subclasses

get_all_subclasses(themes.ThemeClass)
```

```python
import gradio as gr


def greet(name, intensity):
    return "Hello, " + name + "!" * int(intensity)
def get_theme(theme_name):
    themes = {
        "default": gr.Theme(),
        "soft": gr.themes.Soft(),
        "monochrome": gr.themes.Monochrome(),
        "base": gr.themes.Base(),
        "glass": gr.themes.Glass(),
    }
    return themes.get(theme_name, gr.Theme())


def build_app(theme):
    with gr.Blocks(theme=theme) as demo:
        gr.Markdown("## Gradio Theme Demo")

        name = gr.Textbox(label="Name")
        intensity = gr.Slider(1, 5, step=1)
        out = gr.Textbox()

        gr.Button("Greet").click(greet, [name, intensity], out)

    return demo


with gr.Blocks() as demo:
    input_text = gr.Textbox(label="input")

    @gr.render(inputs=input_text)
    def show_split(text):
        # build_app(text)
        gr.Markdown(f"## Gradio Theme Demo{text}")
        gr.Markdown(f"## Gradio Theme Demo{get_theme(text).name}")
        with gr.Row():
            name = gr.Textbox(label="Name")
            intensity = gr.Slider(1, 5, step=1)
            out = gr.Textbox()

            gr.Button("Greet").click(greet, [name, intensity], out)


demo.launch()
```
